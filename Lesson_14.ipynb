## Exercise 2 - Python lists vs NumPy arrays

### Task 1 - Try it with Python Lists

```python
sales_timothy = [10, 15, 20]
sales_brenda = [5, 10, 15]

print(sales_timothy + sales_brenda)
```

1. Python did not produce [15, 25, 35]
2. From the result above, Python concatenated the two lists.
3. This happened because the + operator combines lists by concatenating them instead of performing element-by-element numerical addition.

### Task 2 - Convert them to NumPy arrays

```python
sales_timothy_array = np.array(sales_timothy)
sales_brenda_array = np.array(sales_brenda)

print(f"Combined Sales: {sales_timothy_array + sales_brenda_array}")
```

### Task 3 - Compare the results

| Approach | Result |
| --- | --- |
| Python lists | [10, 15, 20, 5, 10, 15]
|
| NumPy arrays | [15, 25, 35]

Numpy is useful because it allows numerical operations to be performed element-by-element. This makes it convinient to compare and calculate corresponding values in numerical datasets.

# 8. Lagos vs Abuja Comparison

```python
import numpy as np

lagos = [27.5, 29.0, 30.5, 28.0, 31.0]
abuja = [25.0, 28.0, 29.5, 30.0, 32.0]
ibadan = [26.5, 28.5, 29.0, 27.5, 30.5]

lagos_array = np.array(lagos)
abuja_array = np.array(abuja)
ibadan_array = np.array(ibadan)

print(f"Lagos Array: {lagos_array}")
print(f"Abuja Array: {abuja_array}")
print(f"Ibadan Array: {ibadan_array}")
```

### Challenge 1
What location recorded the highest temperature on each day?

```python
highest_temperature_each_day = np.maximum(lagos_array, abuja_array)

highest_temperature_each_day = np.maximum(highest_temperature_each_day, abuja_array)

print(f"Highest Temperature Each Day: {highest_temperature_each_day}")
```

### Challenge 2
Create a calculation that increases every recorded temperature by 1

```python
lagos_temperature_increase = lagos_array + 1
abuja_temperature_increase = abuja_array + 1
ibadan_temperature_increase = ibadan_array + 1

print(f"Lagos: {lagos_temperature_increase}")
print(f"Abuja: {abuja_temperature_increase}")
print(f"Ibadan: {ibadan_temperature_increase}")
```

### Challenge 4
Explain why NumPy becomes increasingly useful as the amount of numerical data grows

Answer:


NumPy becomes increasingly useful because it allows us to store many numerical values in arrays and perform calculations on them efficiently.

Instead of calculating each value seperately, NumPy allows us to perform operations on an entire array element by element.

This makes it easier to work with larger amounts of numerical data.
